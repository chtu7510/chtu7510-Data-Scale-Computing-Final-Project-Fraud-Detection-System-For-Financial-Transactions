```python
# Merge train
train_merged = pd.merge(
    train_transaction, 
    train_identity, 
    on='TransactionID', 
    how='left'  # left join keeps all transaction rows
)

# Merge test
test_merged = pd.merge(
    test_transaction, 
    test_identity, 
    on='TransactionID', 
    how='left'
)

# Quick check
print("Merged train:", train_merged.shape)
print("Merged test:", test_merged.shape)
```

```python
import numpy as np

# Credit Card amounts
cc_amounts = cc['Amount'].values

# IEEE-CIS amounts
train_amounts = train_transaction['TransactionAmt'].values
test_amounts = test_transaction['TransactionAmt'].values
```

```python
cc_fraud_ratio = cc['Class'].mean()  # 0 = normal, 1 = fraud
train_fraud_ratio = train_transaction['isFraud'].mean()

overall_fraud_ratio = (cc_fraud_ratio + train_fraud_ratio) / 2

print("CreditCard fraud ratio:", cc_fraud_ratio)
print("IEEE-CIS train fraud ratio:", train_fraud_ratio)
print("Overall fraud ratio for synthetic generation:", overall_fraud_ratio)
```

```python
# Accounts can be card1
account_list = train_merged['card1'].dropna().unique().tolist()

# Merchants can be addr1 or addr2
merchant_list = train_merged['addr1'].dropna().unique().tolist()

# Optionally: device info or other categorical features
device_list = train_merged['DeviceType'].dropna().unique().tolist()

# Locations (optional)
location_list = train_transaction[['addr1', 'addr2']].dropna().values.tolist()
```

```python
import uuid
import random
import time

def generate_transaction():
    transaction = {
        "TransactionID": str(uuid.uuid4()),
        "TransactionAmt": float(np.random.choice(train_amounts)),
        "isFraud": int(np.random.rand() < overall_fraud_ratio),
        "card1": int(np.random.choice(account_list)),
        "addr1": int(np.random.choice(merchant_list)),
        "DeviceType": str(np.random.choice(device_list)),
        "addr2": int(np.random.choice([loc[1] for loc in location_list]))
    }
    return transaction

# Example: generate 5 synthetic transactions
for _ in range(5):
    print(generate_transaction())
```

```python
import time
import json

def stream_transactions(tps=1000):
    interval = 1 / tps  # seconds per transaction
    while True:
        tx = generate_transaction()
        msg = json.dumps(tx)
        print(msg)  # for now, just print; later send via TCP/UDP
        time.sleep(interval)

# Start streaming at 1000 TPS
# Warning: printing 1000 lines/sec will slow down; reduce for testing
# stream_transactions(tps=10)
```
